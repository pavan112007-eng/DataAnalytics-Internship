```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Fare"], bins=30, kde=True)
plt.title("Fare Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y=df["Age"])
plt.title("Age Boxplot")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y=df["Fare"])
plt.title("Fare Boxplot")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Age", y="Fare", hue="Survived", data=df)
plt.title("Age vs Fare")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['number'])
plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    df[['Survived','Age','Fare','Pclass']])
plt.show()
```

SUMMARY OF FINDINGS

1. Female passengers had significantly higher survival rates.

2. First-class passengers survived more often than lower-class passengers.

3. Fare has a positive relationship with survival.

4. Most passengers were between 20 and 40 years old.

5. The Fare variable is highly skewed and contains outliers.

6. Cabin has a large number of missing values.

7. Passenger class strongly influenced survival probability.
